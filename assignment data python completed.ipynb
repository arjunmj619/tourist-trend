```python
#Subplots — Multiple Visualizations Together
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Cost distribution
axes[0, 0].hist(
    df["Average Cost (USD)"],
    bins=10,
    edgecolor="black"
)
axes[0, 0].set_title("Distribution of Average Cost")
axes[0, 0].set_xlabel("Average Cost (USD)")
axes[0, 0].set_ylabel("Frequency")

# 2. Visitor rating distribution
axes[0, 1].hist(
    df["Visitor Rating"],
    bins=10,
    edgecolor="black"
)
axes[0, 1].set_title("Distribution of Visitor Ratings")
axes[0, 1].set_xlabel("Visitor Rating")
axes[0, 1].set_ylabel("Frequency")

# 3. Cost vs Rating
axes[1, 0].scatter(
    df["Average Cost (USD)"],
    df["Visitor Rating"]
)
axes[1, 0].set_title("Cost vs Visitor Rating")
axes[1, 0].set_xlabel("Average Cost (USD)")
axes[1, 0].set_ylabel("Visitor Rating")

# 4. Visitors vs Rating
axes[1, 1].scatter(
    df["Annual Visitors"],
    df["Visitor Rating"]
)
axes[1, 1].set_title("Annual Visitors vs Visitor Rating")
axes[1, 1].set_xlabel("Annual Visitors")
axes[1, 1].set_ylabel("Visitor Rating")

plt.suptitle("Tourism Dataset — Exploratory Visualizations", fontsize=16)

plt.tight_layout()
plt.show()
```

## Insight Generation and Report

## 1. Key Insights

The exploratory data analysis of the Tourist Trend Dataset revealed several important patterns related to destination cost, visitor numbers, visitor ratings, and tourism seasons.

### Destination Cost

* The dataset contains destinations with different average travel costs.
* The derived **Cost Category** column classifies destinations into Low, Medium, and High cost categories.
* This classification helps compare destinations based on their estimated travel expenses.

### Visitor Trends

* Annual visitor numbers vary considerably between destinations.
* Some destinations attract substantially more visitors than others.
* Grouping the data by country helps identify countries with higher overall tourism demand.

### Visitor Ratings

* Visitor ratings provide an indication of tourist satisfaction with different destinations.
* Destinations with higher ratings can be considered more attractive from a visitor-experience perspective.
* Comparing ratings with cost and annual visitors helps identify whether popularity or cost is associated with visitor satisfaction.

### Seasonal Patterns

* The `Best Season to Visit` column shows that destinations are associated with different preferred travel seasons.
* The distribution of seasons can help identify periods when tourism demand may be more concentrated.

## 2. Correlation Analysis

Correlation analysis was performed on:

* Average Cost (USD)
* Annual Visitors
* Visitor Rating

The correlation matrix helps identify relationships between these numerical variables.

A positive correlation indicates that two variables tend to increase together, while a negative correlation indicates that one tends to decrease as the other increases. However, correlation does not necessarily imply causation.

## 3. Significant Patterns and Anomalies

The visualizations and statistical summaries can be used to identify:

* Destinations with unusually high or low travel costs.
* Destinations receiving exceptionally high numbers of annual visitors.
* Differences in visitor ratings between destinations.
* Possible outliers in cost or annual visitor values.
* Countries that perform differently in terms of tourism cost, popularity, and ratings.

The box plot is particularly useful for identifying potential outliers in average travel costs.

## 4. Data Quality Findings

The dataset was checked during preprocessing.

* Missing values: **None**
* Duplicate rows: **None**
* Data types: **Appropriate for analysis**
* Derived column: **Cost Category**

Therefore, the dataset was suitable for exploratory analysis without requiring extensive data-cleaning operations.

## 5. Overall Conclusion

Overall, the analysis provides a useful overview of tourism trends across different destinations and countries. Travel costs, annual visitor numbers, visitor ratings, and preferred travel seasons show differences between destinations.

The analysis demonstrates that tourism popularity cannot be evaluated using a single factor. A destination may attract many visitors because of its attractions, accessibility, reputation, affordability, or other factors. Similarly, a high visitor rating does not necessarily mean that a destination has the highest number of visitors.

## 6. Recommendations and Next Steps

Based on the analysis, the following recommendations can be considered:

1. **Identify high-performing destinations** based on annual visitors and visitor ratings.
2. **Compare cost and popularity** to identify destinations that provide potentially better value for tourists.

4. **Investigate high-cost destinations** to determine whether higher costs are associated with better visitor ratings.
